```sql
CREATE OR REPLACE TABLE fk_ce_2026_07_1372.gold.dim_date AS
SELECT
    CAST(date_format(d, 'yyyyMMdd') AS INT) AS date_key,
    d AS calendar_date,
    date_format(d, 'EEEE') AS day_name,
    weekofyear(d) AS iso_week,
    CASE WHEN dayofweek(d) IN (1, 7) THEN TRUE ELSE FALSE END AS is_weekend
FROM (
    SELECT explode(sequence(date('2026-03-01'), date('2026-03-31'), interval 1 day)) AS d
);
```

```sql
create or replace table fk_ce_2026_07_1372.gold.dim_store as
select * from fk_ce_2026_07_1372.silver.stores
```

```sql
create or replace table fk_ce_2026_07_1372.gold.dim_product as
select * except (tags, last_changed_snapshot_date)
from fk_ce_2026_07_1372.silver.products
```

```sql
CREATE OR REPLACE TABLE fk_ce_2026_07_1372.gold.dim_customer AS
SELECT
    row_number() OVER (ORDER BY (SELECT NULL)) AS customer_sk,
    *
FROM fk_ce_2026_07_1372.silver.customers;

ALTER TABLE fk_ce_2026_07_1372.gold.dim_customer
    ALTER COLUMN customer_sk SET NOT NULL;

ALTER TABLE fk_ce_2026_07_1372.gold.dim_customer
    ADD CONSTRAINT uc_customer_sk UNIQUE (customer_sk);
```
